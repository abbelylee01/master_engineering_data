```python
# Hashing helpers for dedup/idempotency
import hashlib
import json

# Data and API libraries
import pandas as pd
import requests

# Ingestion metadata helpers
from uuid import uuid4
from datetime import datetime, timezone
```

```python
# API endpoint (increase results for bigger batches)
url = "https://randomuser.me/api/?results=1"

# Batch-level metadata shared across all rows in this pull
batch_id = str(uuid4())
ingest_time = datetime.now(timezone.utc)
```

```python
# Call API and fail fast on HTTP errors
response = requests.get(url, timeout=10)
response.raise_for_status()

# Parse API response JSON
data = response.json()

# Keep raw records first so we can hash the full source object
records = data["results"]

# Flatten nested JSON records into a tabular DataFrame
df = pd.json_normalize(records)

# Raw payload hash (full JSON object, canonicalized for stable hashing)
df["raw_record_hash"] = [
    hashlib.sha256(
        json.dumps(record, sort_keys=True, separators=(",", ":"), ensure_ascii=False).encode("utf-8")
    ).hexdigest()
    for record in records
]

# Add ingestion metadata (same values for all rows in this batch)
df["batch_id"] = batch_id
df["ingest_time"] = ingest_time

# Optional exact-payload deduplication at raw layer
df = df.drop_duplicates(subset=["raw_record_hash"]).reset_index(drop=True)

# Preview raw/landing dataframe
df.head(5)
```

```python
# Select curated columns for the downstream/business view
hash_col = "raw_record_hash" if "raw_record_hash" in df.columns else "record_hash"

selected_df = df[
    [
        "name.first",
        "name.last",
        "phone",
        "gender",
        "registered.age",
        hash_col,
        "batch_id",
        "ingest_time",
        "location.country",
    ]
].copy()

# Standardize column names for readability/consistency
selected_df = selected_df.rename(
    columns={
        "location.country": "country",
        hash_col: "raw_record_hash",
    }
)

# Business-field hash (fast change detection on curated attributes)
business_cols = [
    "name.first",
    "name.last",
    "phone",
    "gender",
    "registered.age",
    "country",
]

selected_df["business_hash"] = (
    pd.util.hash_pandas_object(
        selected_df[business_cols].fillna(""),
        index=False
    )
    .astype("uint64")
    .map(lambda x: f"{x:016x}")
)

# Optional deduplication on business meaning (not raw payload identity)
selected_df = selected_df.drop_duplicates(subset=["business_hash"]).reset_index(drop=True)

# Preview curated dataframe
selected_df
```
